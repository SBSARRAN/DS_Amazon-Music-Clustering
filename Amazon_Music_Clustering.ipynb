```python
num_cols = ['duration_ms','danceability','energy','loudness','popularity_songs',
            'speechiness','acousticness','instrumentalness','liveness','valence','tempo']
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(15, 12))
for i, col in enumerate(num_cols, 1):
    plt.subplot(5, 3, i)
    sns.boxplot(data=df, y=col)
    plt.title(f'{col} (skewness : {df[col].skew():.3f})')
plt.suptitle('Features Before Processing', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
for i in num_cols:
    if (df[i] > 0).all():
        if df[i].skew() > 1:
            df[i] = np.log(df[i])
    elif df[i].skew() < -1:
        df[i] = np.power(df[i], 2)
```

```python
# After applying log to skewness
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(15, 12))
for i, col in enumerate(num_cols, 1):
    plt.subplot(5, 3, i)
    sns.boxplot(data=df, y=col)
    plt.title(f'{col} (skewness : {df[col].skew():.3f})')
plt.suptitle('Features after Processing', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler
X = df[num_cols].copy()
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df)
df_scaled 
```

```python
# Find the best K using Elbow Method
from sklearn.cluster import KMeans
inertia = []
K_value = range(2, 11)  

for k in K_value:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(df_scaled)
    inertia.append(kmeans.inertia_)
    print('K value = ', k, 'Inertia value = ', kmeans.inertia_)

plt.plot(K_value, inertia, 'bx-')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Inertia (SSE)')
plt.title('Elbow Method For Optimal k')
plt.show()
```

```python
# Evaluate the silhouette score
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt
scores = []
for k in range(2,11):
    kmeans = KMeans(n_clusters=k,random_state=42)
    kmeans.fit(df_scaled)
    score = silhouette_score(df_scaled,kmeans.labels_)
    scores.append(score)
plt.plot(range(2,11),scores,'bx-')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Method for Optimal k')
plt.show()
    
```

```python
# silhouette score from K values
scores
```

```python
# After Elbow Methot and Silhouette Method the best K = 3
# So the best K value is 3
kmeans = KMeans(n_clusters=3,random_state=42)
kmeans.fit(df_scaled)
df['cluster_KMEANS'] = kmeans.labels_
```

```python
# Evaluate the K_Means
from sklearn.metrics import silhouette_score
sil_score = silhouette_score(df_scaled,df['cluster_KMEANS'])
print(f'silhouette_score = {sil_score}')
```

```python
cluster_summary = df.groupby('cluster_KMEANS').mean()
cluster_summary
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
pca_data = pca.fit_transform(df_scaled)
centroids = pca.transform(kmeans.cluster_centers_)

plt.figure(figsize=(8,6))
plt.scatter(pca_data[:,0], pca_data[:,1], c=df['cluster_KMEANS'], cmap='viridis', alpha=0.6)
plt.scatter(centroids[:,0], centroids[:,1], c='red', s=200, marker='X', label='Centroids')
plt.title('Clusters Visualized using PCA with Centroids')
plt.xlabel('PCA 1')
plt.ylabel('PCA 2')
plt.legend()
plt.show()
```
